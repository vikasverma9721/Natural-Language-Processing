```python
# Read & print only the first line
with open('test.txt', 'r') as txt:
  first_line = txt.readlines()[0]
print(first_line)
```

```python
# Read & print file line by line
with open('test.txt', 'r') as txt:
  for line in txt:
    print(line,end='')
```
